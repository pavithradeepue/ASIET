```python
# ------------------------------------------------------------
# ASSIGNMENT 1: NLP TEXT PREPROCESSING
# Name: Pavithra Deepu E
# ------------------------------------------------------------

import re
import nltk

from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords, wordnet
from nltk.stem import PorterStemmer
from nltk.stem import WordNetLemmatizer
from nltk import pos_tag

# Download the required NLTK resources
nltk.download("punkt", quiet=True)
nltk.download("punkt_tab", quiet=True)
nltk.download("stopwords", quiet=True)
nltk.download("wordnet", quiet=True)
nltk.download("omw-1.4", quiet=True)
nltk.download("averaged_perceptron_tagger", quiet=True)
nltk.download("averaged_perceptron_tagger_eng", quiet=True)

# Original self-introduction
original_text = """
My name is Pavithra Deepu E, and I am from Kochi, Kerala.
I completed my schooling at Jyothi Nivas Public School, where I secured 87.8% in CBSE
Class 12. I come from a supportive family that continuously encourages my education,
ambitions, and personal growth. My father is a technician at AVT McCormick Pvt Ltd,
my mother is a former Computer Science teacher, and my younger brother is currently
studying in 9th standard. I am currently a final-year B.Tech student in Computer Science
and Engineering with Artificial Intelligence at Adi Shankara Institute of Engineering and
Technology, with a CGPA of 8.54. I am also currently pursuing a six-month internship
at the Centre for Artificial Intelligence and Robotics (CAIR),
Defence Research and Development Organisation (DRDO).

My achievements include co-authoring an IEEE publication on the early
detection of plant diseases using deep learning, winning first prize
at THREADS'26 for the Thrift and Go sustainable fashion platform,
securing first runner-up at Girlathon'25 for the V.I.R.A emergency
safety wearable, and winning third prize for the MoodDecode NLP-based
emotion and crisis-detection platform.

I have also held leadership positions such as IEEE Women in Engineering
Affinity Group Chairperson, IEEE RAS Kerala Chapter Women in Engineering
Coordinator, and Women Entrepreneurship Lead at IEDC. My hobbies include
dancing and painting. I am particularly interested in artificial
intelligence, machine learning, natural language processing, computer
vision, generative AI, and developing technologies that create a
positive social impact. My strengths include leadership, communication,
teamwork, problem-solving, public speaking, and time management. One of
my weaknesses is that I sometimes take on several responsibilities at
the same time, but I am learning to prioritise my tasks and manage my
workload more effectively.
"""

# Step 1: Convert the text to lowercase
lowercase_text = original_text.lower()

# Step 2: Remove URLs
text_without_urls = re.sub(
    r"http\S+|https\S+|www\S+",
    "",
    lowercase_text
)

# Step 3: Remove HTML tags
text_without_html = re.sub(
    r"<.*?>",
    "",
    text_without_urls
)

# Step 4: Remove numbers
text_without_numbers = re.sub(
    r"\d+(\.\d+)?",
    "",
    text_without_html
)

# Step 5: Remove punctuation and special characters
text_without_punctuation = re.sub(
    r"[^a-zA-Z\s]",
    " ",
    text_without_numbers
)

# Step 6: Remove extra whitespace
clean_text = re.sub(
    r"\s+",
    " ",
    text_without_punctuation
).strip()

# Step 7: Tokenisation
tokens = word_tokenize(clean_text)

# Step 8: Stop-word removal
english_stopwords = set(stopwords.words("english"))

filtered_tokens = [
    word
    for word in tokens
    if word not in english_stopwords
]

# Step 9: Stemming
stemmer = PorterStemmer()

stemmed_tokens = [
    stemmer.stem(word)
    for word in filtered_tokens
]

# Convert NLTK POS tags into WordNet POS tags
def get_wordnet_pos(tag):
    if tag.startswith("J"):
        return wordnet.ADJ
    elif tag.startswith("V"):
        return wordnet.VERB
    elif tag.startswith("N"):
        return wordnet.NOUN
    elif tag.startswith("R"):
        return wordnet.ADV
    else:
        return wordnet.NOUN

# Step 10: POS-aware lemmatisation
lemmatizer = WordNetLemmatizer()
tagged_tokens = pos_tag(filtered_tokens)

lemmatized_tokens = [
    lemmatizer.lemmatize(
        word,
        get_wordnet_pos(tag)
    )
    for word, tag in tagged_tokens
]

# Display all the preprocessing outputs
print("=" * 70)
print("1. ORIGINAL TEXT")
print("=" * 70)
print(original_text)

print("=" * 70)
print("2. LOWERCASE TEXT")
print("=" * 70)
print(lowercase_text)

print("=" * 70)
print("3. TEXT AFTER REMOVING URLs AND HTML TAGS")
print("=" * 70)
print(text_without_html)

print("=" * 70)
print("4. TEXT AFTER REMOVING NUMBERS")
print("=" * 70)
print(text_without_numbers)

print("=" * 70)
print("5. TEXT AFTER REMOVING PUNCTUATION")
print("=" * 70)
print(text_without_punctuation)

print("=" * 70)
print("6. CLEANED TEXT")
print("=" * 70)
print(clean_text)

print("=" * 70)
print("7. TOKENISED TEXT")
print("=" * 70)
print(tokens)

print("=" * 70)
print("8. TOKENS AFTER STOP-WORD REMOVAL")
print("=" * 70)
print(filtered_tokens)

print("=" * 70)
print("9. STEMMED TOKENS")
print("=" * 70)
print(stemmed_tokens)

print("=" * 70)
print("10. POS-TAGGED TOKENS")
print("=" * 70)
print(tagged_tokens)

print("=" * 70)
print("11. LEMMATISED TOKENS")
print("=" * 70)
print(lemmatized_tokens)

print("=" * 70)
print("FINAL PREPROCESSED TEXT")
print("=" * 70)
print(" ".join(lemmatized_tokens))

print("\nOriginal token count:", len(tokens))
print("Filtered token count:", len(filtered_tokens))
print("\nProgram executed successfully.")
```
